## Project Problem Statement:

The aviation industry experiences operational challenges due to increasing passenger traffic, airport congestion, weather conditions, ATC restrictions, and limited ground resources. These issues lead to flight delays, inefficient gate usage, and longer aircraft turnaround times. This project analyzes a realistic dataset of 20,000 flight records using Python and Power BI to identify operational bottlenecks, measure airport performance, and generate data-driven insights without using machine learning.

### Success Metrics :

The project will be successful if it can:

Clean and validate the dataset.
Calculate key airport performance KPIs.
Identify the main causes of flight delays.
Measure airport congestion and gate utilization.
Evaluate aircraft turnaround efficiency.
Analyze passenger traffic across airlines and airports.
Generate actionable operational insights.
Develop an interactive Power BI dashboard.
Document the complete data analytics workflow and results.

### Business Meaning of the 35 Dataset Columns

Column                           Business Meaning                                           

Flight_ID                    : Unique identifier for every flight.                  
Airline                      : Airline operating the flight.                              
Origin                       : Airport from which the flight departs.                     
Destination                  : Airport where the flight arrives.                          
Aircraft_Type                : Type/model of aircraft used for the flight.                
Terminal                     : Airport terminal assigned to the flight.                   
Gate                         : Gate assigned for boarding and arrival.                    
Weather_Condition            : Weather affecting airport operations.                      
Flight_Date                  : Date of flight operation.                                  
Month                        : Month of operation for trend analysis.                     
Scheduled_Departure          : Planned departure time.                                    
Scheduled_Arrival            : Planned arrival time.                                      
Status                       : Current flight status (On Time, Delayed, Cancelled, etc.). 
Gate_Delay_Min               : Delay caused by unavailable or occupied gates.             
Fuel_Delay_Min               : Delay due to aircraft refueling operations.                
Baggage_Delay_Min            : Delay caused by baggage loading or unloading.              
Catering_Delay_Min           : Delay due to catering services.

Cleaning_Delay_Min           : Delay due to aircraft cleaning activities.                 
Security_Delay_Min           : Delay resulting from security procedures.                  
Maintenance_Delay_Min        : Delay due to aircraft maintenance.                         
ATC_Delay_Min                : Delay imposed by Air Traffic Control.                      
Total_Delay_Min              : Overall operational delay from all contributing factors.   
Departure_Delay_Min          : Difference between scheduled and actual departure time.    
Arrival_Delay_Min            : Difference between scheduled and actual arrival time.      
Passenger_Count              : Number of passengers onboard.                              
Load_Factor (%)              : Percentage of available seats occupied by passengers.      
Ground_Staff_Count           : Ground personnel assigned to service the flight.           
Baggage_Count                : Total checked baggage handled for the flight.              
Fuel_Quantity_Liters         : Fuel loaded into the aircraft before departure.            
Gate_Occupancy_Min           : Duration the aircraft occupies the assigned gate.          
Turnaround_Time_Min          : Time between aircraft arrival and subsequent departure.    
Actual_Departure             : Actual departure timestamp.                                
Actual_Arrival               : Actual arrival timestamp.                                  
Delay_Reason                 : Primary reason responsible for the flight delay.
