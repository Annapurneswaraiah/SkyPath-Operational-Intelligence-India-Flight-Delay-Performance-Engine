```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=pd.read_csv(r"C:\Users\asus\Python class\Intership\Airline_cleaned_dataset13.csv")
```

```python
df.head()
```

```python
cols = ["Baggage_Delay_Min",
        "Total_Delay_Min",
        "Passenger_Count"]

print(df[cols].describe())
```

```python
cols = ["Baggage_Delay_Min",
        "Total_Delay_Min",
        "Passenger_Count"]

for col in cols:
    plt.figure(figsize=(7,4))
    plt.hist(df[col], bins=30, edgecolor="black")
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
for col in cols:
    print(col)
    print("Skewness:", df[col].skew())
    print()
```

```python
for col in cols:
    print(col)
    print("Kurtosis:", df[col].kurt())
    print()
```

```python
summary = pd.DataFrame({
    "Mean": df[cols].mean(),
    "Median": df[cols].median(),
    "Std": df[cols].std(),
    "Skewness": df[cols].skew(),
    "Kurtosis": df[cols].kurt()
})

print(summary)
```

- The dataset contains approximately 20,000 airport operation records.
- The analysis depends entirely on the available sample and may not represent every airport or airline.
- Missing values were imputed during the cleaning process, which may slightly affect summary statistics.
- Extreme delay values (outliers) can influence the mean and standard deviation.
- The dataset represents historical operations and may not generalize to future flight patterns or unusual events.
