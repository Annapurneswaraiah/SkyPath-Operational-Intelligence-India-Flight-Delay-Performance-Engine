```python
import pandas as pd

# Load the dataset
import pandas as pd

df = pd.read_excel(r"C:\Users\asus\Downloads\Airport_Operations_Dataset_Perfect_Cleaned.xlsx")

# Count rows and columns
rows, columns = df.shape

print("Number of Rows:", rows)
print("Number of Columns:", columns)
df.head(2)
```

```python
print(df.dtypes)
```

```python
missing_values = df.isnull().sum()
print(missing_values)
```

```python
duplicates = df[df.duplicated(subset=["Flight_ID"], keep=False)]
```

```python
print(duplicates)
```

```python
print("Total Duplicate Records:", duplicates.shape[0])
```

```python

df['Scheduled_Arrival'] = pd.to_datetime(df['Scheduled_Arrival'])
df['Actual_Arrival'] = pd.to_datetime(df['Actual_Arrival'])

df['Actual_Arrival'] = df['Actual_Arrival'].fillna(
    df['Scheduled_Arrival'] + pd.to_timedelta(df['Arrival_Delay_Min'], unit='m')
)

df['Scheduled_Departure'] = pd.to_datetime(df['Scheduled_Departure'])
df['Actual_Departure'] = pd.to_datetime(df['Actual_Departure'])

df['Actual_Departure'] = df['Actual_Departure'].fillna(
    df['Scheduled_Departure'] + pd.to_timedelta(df['Departure_Delay_Min'], unit='m')
)


print("Missing values after filling:")
print(df[['Actual_Arrival', 'Actual_Departure']].isnull().sum())
```

```python
missing_values = df.isnull().sum()
print(missing_values)
```

```python

df.to_csv("Airport_Operations_Dataset_Cleaned.csv", index=False)

print("File successfully saved as CSV!")
```
