- EDA on Baggage_Delay_Min, Total_Delay_Min, Passenger_Count — distributions, summary stats
- Segment analysis by Aircraft_Type, Status
- Derive date parts from Flight_Date, Month

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"C:\Users\asus\Python class\Intership\Airport_Operations_Dataset_Cleaned5.csv")
```

```python
df.head()
```

```python
num_cols=['Baggage_Delay_Min', 'Total_Delay_Min', 'Passenger_Count']
```

```python
df[num_cols].describe()
```

```python
for col in num_cols:
    sns.histplot(df[col],kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
df.groupby('Aircraft_Type')[num_cols].mean()
```

```python
df.groupby('Status')[num_cols].mean()
```

```python
sns.boxplot(x='Aircraft_Type',y='Total_Delay_Min',data=df)
```

```python
sns.barplot(x='Status',y='Passenger_Count',data=df)
```

```python
sns.lineplot(x='Month',y='Total_Delay_Min',data=df)
```

```python
sns.countplot(x="Day",data=df)
```
