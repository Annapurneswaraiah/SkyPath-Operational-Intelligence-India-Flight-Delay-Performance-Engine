- Open Airport_Operations_20k_Realistic.csv in Excel; apply filters and conditional formatting
- Apply cleaning plan in Excel: fix types, trim whitespace, harmonise Aircraft_Type, Status
- Validate and reconcile row counts against original file

```python
import pandas as pd
```

```python
df=pd.read_csv(r"C:\Users\asus\Python class\Intership\Airport_Operations_Dataset_Cleaned5.csv")
```

```python
df.info()
```

```python
original_rows = df.shape[0]
print("Original Rows:", original_rows)
```

```python
df["Aircraft_Type"] = df["Aircraft_Type"].str.strip()
df["Status"] = df["Status"].str.strip()
```

```python
df["Aircraft_Type"] = df["Aircraft_Type"].replace({
    "A320 ": "A320",
    "A 320": "A320",
    "B737 ": "B737"})
```

```python
df["Status"] = df["Status"].replace({
    "Delayed ": "Delayed",
    " delayed": "Delayed",
    "On Time": "On-Time",
    "On time": "On-Time",
    "Cancelled ": "Cancelled"
})
```

```python
df["Flight_Date"] = pd.to_datetime(df["Flight_Date"], errors="coerce")
```

```python
cleaned_rows = df.shape[0]

print("Cleaned Rows:", cleaned_rows)
```

```python
if original_rows == cleaned_rows:
    print("Row count matched. No rows lost.")
else:
    print("Row count mismatch.")
```
