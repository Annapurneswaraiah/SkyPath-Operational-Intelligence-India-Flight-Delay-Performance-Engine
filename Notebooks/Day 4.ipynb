```python
import pandas as pd
```

```python

df = pd.read_csv(r"C:\Users\asus\Downloads\Airport_Operations_20k_Realistic.csv")

df["Flight_ID"] = df["Flight_ID"].astype(str)
print("Data Types:")
print(df.dtypes)

for col in df.columns:
    if df[col].dtype == 'object':
        try:
            pd.to_numeric(df[col])
            print(f"{col} may contain numeric values stored as text.")
        except:
            pass

df["Flight_Date"] = pd.to_datetime(df["Flight_Date"],errors="coerce")

invalid_dates = df[df["Flight_Date"].isnull()]

print("\nInvalid Flight_Date Records:")
print(invalid_dates[["Flight_Date"]])

# Check Month column
print("\nUnique Month Values:")
print(df["Month"].unique())
```

```python
invalid_baggage = df[df["Baggage_Delay_Min"] < 0]

invalid_total = df[df["Total_Delay_Min"] < 0]


invalid_passenger = df[
    (df["Passenger_Count"] <= 0) |
    (df["Passenger_Count"] > 850)
]

print("Invalid Baggage Delay:", len(invalid_baggage))
print("Invalid Total Delay:", len(invalid_total))
print("Invalid Passenger Count:", len(invalid_passenger))
```

```python
print("Aircraft Types")
print(df["Aircraft_Type"].value_counts())

print("\nStatus Values")
print(df["Status"].value_counts())
```

```python

report = []


report.append("Missing Values")
report.append(str(df.isnull().sum()))
report.append("\n")


duplicates = df.duplicated(subset="Flight_ID").sum()
report.append(f"Duplicate Flight_IDs : {duplicates}")
report.append("\n")

report.append("Data Types")
report.append(str(df.dtypes))
report.append("\n")


invalid_baggage = len(df[df["Baggage_Delay_Min"] < 0])
invalid_total = len(df[df["Total_Delay_Min"] < 0])
invalid_passenger = len(df[(df["Passenger_Count"] <= 0) | (df["Passenger_Count"] > 850)])

report.append(f"Invalid Baggage Delay : {invalid_baggage}")
report.append(f"Invalid Total Delay : {invalid_total}")
report.append(f"Invalid Passenger Count : {invalid_passenger}")
report.append("\n")

report.append("Aircraft Types")
report.append(str(df["Aircraft_Type"].value_counts()))

report.append("\nStatus")
report.append(str(df["Status"].value_counts()))


with open("Data_Profiling_Report.txt", "w") as file:
    file.write("\n".join(report))

print("Data Profiling Report Created Successfully.")
```
